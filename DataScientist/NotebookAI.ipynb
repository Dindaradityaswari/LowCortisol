## Cleaning Data

```python
# hapus baris duplikat
df = df.drop_duplicates()
print(f"Jumlah baris duplikat: {df.duplicated().sum()}")
```

```python
# drop kolom tidak dipakai
df = df.drop(columns=['employee_id', 'nama', 'tanggal_bergabung', 'tanggal_isi_survei', 'deskripsi_keluhan'])
df.head()
```

```python
# menghapus baris yang tingkat burnout nya null
df["tingkat_burnout"] = pd.to_numeric(df["tingkat_burnout"], errors="coerce")
df = df.dropna(subset=["tingkat_burnout"])

df = df.reset_index(drop=True)
```

```python
#mengganti noise menjadi nan
NOISE = {"-", "?", "unknown", "Unknown", "tidak tahu", "Tidak Tahu",
         "tidak mau menyebutkan", "Tidak Mau Menyebutkan", ""}

def to_nan(series):
    """Ganti nilai noise → NaN."""
    return series.apply(lambda x: np.nan if str(x).strip() in NOISE else x)
```

```python
# standarisasi kolom jenis kelamin
def std_gender(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"l","laki","laki-laki","pria","male","cowo","lk"}: return "Laki-laki"
    if v in {"p","perempuan","wanita","female","cewe","pr"}:    return "Perempuan"
    return np.nan
df["jenis_kelamin"] = to_nan(df["jenis_kelamin"]).apply(std_gender)
```

```python
# standarisasi kolom pendidikan terakhir
def std_edu(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower().replace("-","").replace(" ","")
    m = {"sd":"SD","smp":"SMP","sma":"SMA","smk":"SMA","smasmk":"SMA",
         "d3":"D3","diploma3":"D3","s1":"S1","sarjana":"S1",
         "s2":"S2","magister":"S2","s3":"S3","doktor":"S3"}
    return m.get(v, np.nan)
df["pendidikan_terakhir"] = to_nan(df["pendidikan_terakhir"]).apply(std_edu)
```

```python
# standarisasi kolom status pernikahan
def std_marital(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"belum menikah","lajang","single"}:                    return "Belum Menikah"
    if v in {"menikah","married"}:                                   return "Menikah"
    if v in {"cerai","divorced","duda","janda","duda/janda"}:        return "Cerai/Duda/Janda"
    return np.nan
df["status_pernikahan"] = to_nan(df["status_pernikahan"]).apply(std_marital)
```

```python
# standarisasi kolom tipe perusahaan
def std_company(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    m = {"swasta":"Swasta","corporate":"Swasta","korporat":"Swasta",
         "bumn":"BUMN","badan usaha milik negara":"BUMN","pma":"PMA",
         "startup":"Startup","pemerintah":"Pemerintah","ngo":"NGO",
         "akademik":"Akademik","freelance":"Freelance",
         "remote":"Remote","wfh":"Remote","lainnya":"Lainnya"}
    return m.get(v, np.nan)
df["tipe_perusahaan"] = to_nan(df["tipe_perusahaan"]).apply(std_company)
```

```python
# standarisasi kolom ukuran perusahaan (menghapus kata orang)
df["ukuran_perusahaan"] = (to_nan(df["ukuran_perusahaan"])
    .str.strip()
    .str.replace(r"\s*[Oo]rang$", "", regex=True)
    .str.strip())
```

```python
# standarisasi kolom rentang gaji
def std_salary(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if any(k in v for k in ["tidak","tdk","mau"]): return np.nan
    import re
    v = re.sub(r"(\d+)-(\d+)\s*jt\b", r"\1-\2 juta", v)
    v = v.replace("jt"," juta").replace("  "," ").strip()
    return v.title()
df["rentang_gaji"] = to_nan(df["rentang_gaji"]).apply(std_salary)
```

```python
# standarisasi kolom status wfh
def std_wfh(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"ya","y","yes","iya"}:                                          return "Ya"
    if v in {"tidak","t","no","nope"}:                                       return "Tidak"
    if v in {"hybrid","campuran","kadang","tergantung","wfh & wfo","wfh&wfo"}: return "Hybrid"
    return np.nan
df["status_wfh"] = to_nan(df["status_wfh"]).apply(std_wfh)
```

```python
# standarisasi kolom jenis shift
def std_shift(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    m = {"pagi":"Pagi","morning":"Pagi","08:00-17:00":"Pagi",
         "siang":"Siang","12:00-21:00":"Siang",
         "malam":"Malam","night shift":"Malam","22:00-06:00":"Malam",
         "fleksibel":"Fleksibel","flexible":"Fleksibel","wfh bebas":"Fleksibel",
         "rotasi":"Rotasi","shifting":"Rotasi"}
    return m.get(v, np.nan)
df["jenis_shift"] = to_nan(df["jenis_shift"]).apply(std_shift)
```

```python
# standarisasi kolom beban kerja persepsi
def std_workload(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"ringan","sangat ringan","normal"}:     return "Ringan"
    if v in {"sedang","medium"}:                     return "Sedang"
    if v in {"berat","heavy","tinggi","overload"}:   return "Berat"
    if v in {"sangat berat","overwhelmed"}:          return "Sangat Berat"
    return np.nan
df["beban_kerja_persepsi"] = to_nan(df["beban_kerja_persepsi"]).apply(std_workload)
```

```python
# standarisasi kolom waktu tidur malam
def std_sleep_time(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if any(k in v for k in ["tidak tentu","tidak tetap","bervariasi"]): return "Tidak Tentu"
    return str(x).strip().title()
df["waktu_tidur_malam"] = to_nan(df["waktu_tidur_malam"]).apply(std_sleep_time)
```

```python
# standarisasi kolom status merokok
def std_smoke(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"ya","yes","iya","merokok"}:        return "Ya"
    if v in {"tidak","no","tidak merokok"}:      return "Tidak"
    if v in {"kadang","social smoker","sosial"}: return "Kadang"
    return np.nan
df["status_merokok"] = to_nan(df["status_merokok"]).apply(std_smoke)
```

```python
# standarisasi kolom pola makan
def std_food(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"buruk","sangat buruk","tidak teratur","jarang makan","sering skip makan"}: return "Buruk"
    if v in {"cukup","sedang"}:                                                          return "Cukup"
    if v in {"baik","teratur","sehat"}:                                                  return "Baik"
    return np.nan
df["pola_makan"] = to_nan(df["pola_makan"]).apply(std_food)
```

```python
# standarisasi akses layanan kesehatan mental
def std_mh_access(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"ya","sedang menjalani","online konseling"}: return "Ya/Aktif"
    if v in {"pernah","sudah pernah","belum pernah"}:     return "Pernah"
    if v in {"tidak","tidak ada akses","tidak mampu biaya","tidak tahu ada"}: return "Tidak"
    return np.nan
df["akses_layanan_kesehatan_mental"] = to_nan(df["akses_layanan_kesehatan_mental"]).apply(std_mh_access)
```

```python
# standarisasi kolom keamanan pekerjaan
def std_job_sec(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    m = {"sangat tidak aman":"Sangat Tidak Aman","tidak aman":"Tidak Aman",
         "khawatir":"Tidak Aman","sangat khawatir":"Sangat Tidak Aman",
         "netral":"Cukup Aman","cukup aman":"Cukup Aman",
         "aman":"Aman","sangat aman":"Sangat Aman"}
    return m.get(v, np.nan)
df["keamanan_pekerjaan"] = to_nan(df["keamanan_pekerjaan"]).apply(std_job_sec)
```

```python
# standarisasi kolom kebijakan mh perusahaan
def std_mh_policy(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"ada","ada dan efektif","ada tapi jarang dipakai","ada tapi tidak efektif","ya"}: return "Ada"
    if v in {"tidak ada","tidak"}:               return "Tidak Ada"
    if v in {"tidak tahu","kurang tahu"}:        return "Tidak Tahu"
    return np.nan
df["kebijakan_mh_perusahaan"] = to_nan(df["kebijakan_mh_perusahaan"]).apply(std_mh_policy)
```

```python
# standarisasi kolom status kesehatan mental
def std_mh_status(x):
    if pd.isna(x): return np.nan
    v = str(x).strip().lower()
    if v in {"buruk","sangat buruk","tidak baik","bad","parah"}: return "Buruk"
    if v in {"cukup","cukup baik","ok","okay","fine"}:           return "Cukup"
    if v in {"baik","good"}:                                      return "Baik"
    return np.nan
df["status_kesehatan_mental"] = to_nan(df["status_kesehatan_mental"]).apply(std_mh_status)
```

```python
# standarisasi kolom departemen, jabatan, keluhan fisik utama, dan riwayat kesehatan mentall
for col in ["departemen","jabatan","keluhan_fisik_utama","riwayat_kesehatan_mental"]:
    df[col] = to_nan(df[col]).str.strip().str.title()
```

```python
# standarisasi kolom numerik
# Batas logis per kolom (min, max)
numeric_bounds = {
    "usia"                          : (15,  70),
    "jumlah_tanggungan"             : (0,   20),
    "lama_bekerja_tahun"            : (0,   50),
    "jam_kerja_per_hari"            : (0,   24),
    "jam_lembur_per_hari"           : (0,   16),
    "hari_kerja_per_minggu"         : (0,    7),
    "frekuensi_meeting_per_hari"    : (0,   20),
    "jumlah_deadline_per_minggu"    : (0,   50),
    "jam_tidur_per_hari"            : (0,   24),
    "kualitas_tidur"                : (0,   10),
    "jam_istirahat_siang"           : (0,    5),
    "frekuensi_olahraga_per_minggu" : (0,   14),
    "durasi_olahraga_menit"         : (0,  300),
    "jam_layar_per_hari"            : (0,   24),
    "konsumsi_kafein_per_hari"      : (0,   20),
    "tingkat_stres"                 : (0,   10),
    "tingkat_kecemasan"             : (0,   10),
    "tingkat_depresi"               : (0,   10),
    "motivasi_kerja"                : (0,   10),
    "kepuasan_kerja"                : (0,   10),
    "work_life_balance"             : (0,   10),
    "produktivitas_diri"            : (0,   10),
    "kemampuan_konsentrasi"         : (0,   10),
    "dukungan_sosial_rekan"         : (0,   10),
    "dukungan_atasan"               : (0,   10),
    "frekuensi_konflik_kerja"       : (0,   10),
    "tingkat_burnout"               : (0,   10),
}

for col, (lo, hi) in numeric_bounds.items():
    df[col] = (
        pd.to_numeric(df[col], errors="coerce")   # konversi, error → NaN
        .clip(lower=lo, upper=hi)                  # buang outlier tidak logis
        .round(0)                                  # bulatkan ke bilangan bulat
        .astype("Int64")                           # simpan sebagai integer (support NaN)
    )

# Pastikan target tetap bersih setelah konversi
df = df.dropna(subset=["tingkat_burnout"]).reset_index(drop=True)
```

```python
# mengatasi missing value untuk kolom numerik (dikonversi menggunakan median)
for col in [c for c in numeric_bounds if c != "tingkat_burnout"]:
    if df[col].isna().any():
        median_val = int(df[col].dropna().median())
        df[col] = df[col].fillna(median_val)
```

```python
# mengatasi missing value untuk kolom kategorikal (dikonversi menggunakan modus)
cat_cols = [
    "jenis_kelamin","pendidikan_terakhir","status_pernikahan",
    "departemen","jabatan","tipe_perusahaan","ukuran_perusahaan",
    "rentang_gaji","status_wfh","jenis_shift","beban_kerja_persepsi",
    "status_merokok","pola_makan","waktu_tidur_malam",
    "keluhan_fisik_utama","riwayat_kesehatan_mental",
    "akses_layanan_kesehatan_mental","keamanan_pekerjaan",
    "kebijakan_mh_perusahaan","status_kesehatan_mental",
]
for col in cat_cols:
    if df[col].isna().any():
        mode_val = df[col].mode()
        if len(mode_val) > 0:
            df[col] = df[col].fillna(mode_val[0])
```

```python
# menyimpan data bersih
df.to_csv("dataAI_cleaned.csv", index=False)
print("\nFile tersimpan: dataAI_cleaned.csv")
print(f"Total baris: {len(df)} | Total kolom: {df.shape[1]}")
```
